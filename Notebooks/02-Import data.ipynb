FUNCIONES BASICAS DE PANDAS 

```python
# Cargar la tabla CGSE como DataFrame de pandas
cgse = spark.table("workspace.electiva.cgse").toPandas()

# Filtrando a nivel de columna
cgse['FEX_C18']

#Aplicando operaciones
cgse['PERIODO'].sum()
cgse['PERIODO'].max()
cgse['PERIODO'].min()
cgse['PERIODO'].mean()
cgse['PERIODO'].std()
cgse['PERIODO'].var()
cgse['PERIODO'].median()
cgse['PERIODO'].mode()
```

```python
cgse.head()
```

```python
# ============================================================
# FUNCIONES BÁSICAS DE PANDAS
# ============================================================

# --- Información general del DataFrame ---
print("Shape:", cgse.shape)
print("Columnas:", list(cgse.columns))
print("\nTipos de dato:")
print(cgse.dtypes)

# --- Inspección de datos ---
print("\nFilas duplicadas:", cgse.duplicated().sum())
print("Nulos por columna:")
print(cgse.isnull().sum())

# --- Selección de columnas y filas ---
print("\nColumna PERIODO:")
print(cgse['PERIODO'].head())
print("\nPrimeras 5 filas con iloc:")
display(cgse.iloc[0:5])

# --- Filtrado ---
print("\nFiltro simple (PERIODO == 20260626):")
display(cgse[cgse['PERIODO'] == 20260626].head())

# --- Operaciones por columna ---
print("\nEstadísticas de PERIODO:")
print(f"  Suma:   {cgse['PERIODO'].sum()}")
print(f"  Media:  {cgse['PERIODO'].mean()}")
print(f"  Mediana:{cgse['PERIODO'].median()}")
print(f"  Mín:    {cgse['PERIODO'].min()}")
print(f"  Máx:    {cgse['PERIODO'].max()}")
print(f"  Std:    {cgse['PERIODO'].std():.2f}")
print(f"  Count:  {cgse['PERIODO'].count()}")

# --- Agrupaciones ---
print("\nAgrupación por PERIODO (conteo):")
print(cgse.groupby('PERIODO').size())

# --- Ordenamiento ---
print("\nOrden descendente por FEX_C18 (top 5):")
display(cgse.sort_values('FEX_C18', ascending=False).head())

# --- Describe general ---
display(cgse.describe())
```
